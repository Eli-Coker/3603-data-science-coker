## 📦 One Branch, Many Lines
A branch isn't limited to one statement. **Everything indented under the `if` (or `else`) runs together as a block** — as many lines as you need, as long as they stay indented at the same level.

````python
cart_total = 84.50
is_member = False

if is_member:
    discount = cart_total * 0.10
    final_total = cart_total - discount
    print(f'Member discount applied: -${discount:.2f}')
    print(f'Final total: ${final_total:.2f}')
else:
    final_total = cart_total
    print('No discount applied.')
    print(f'Final total: ${final_total:.2f}')
#If is_member were False, my guess is that 2 lines would print, with the second line saying "Final total: $84.50" 
````

## 🔀 Multiple Branches

````python
temp = 72
if temp > 90:
    print('Too hot!')
elif temp < 60:
    print('Too cold!')
else:
    print('Just right.')
````

Just like a single `if`/`else`, each `elif`/`else` branch in a chain can hold multiple lines too:

````python
score = 65

if score >= 90:
    letter = 'A'
    feedback = 'Excellent work!'
elif score >= 80:
    letter = 'B'
    feedback = 'Great job!'
elif score >= 70:
    letter = 'C'
    feedback = 'Solid effort, keep practicing.'
else:
    letter = 'F'
    feedback = "Let's review this together."

print(f'Grade: {letter}')
print(feedback)

#My guess is that if the score is 65, line 1 will print "Grade: F" and line 2 will print "Let's review this together."
````

## 🔗 Logical Operators
Combine multiple conditions with `and`, `or`, and `not`.

````python
age = 20
has_id = True
print(age >= 18 and has_id)
print(age < 13 or age > 65)
print(not has_id)
````

### 🔀 Cross-Language Note: C++ Logical Operators
If you've written C++, this is the same idea with different spelling — Python spells them out, C++ uses symbols:

| Python | C++ |
|--------|-----|
| `and`  | `&&` |
| `or`   | `\|\|` |
| `not`  | `!` |

```cpp
// C++
bool age_ok = age >= 18 && has_id;
bool out_of_range = age < 13 || age > 65;
bool no_id = !has_id;
```
Behavior-wise they work the same way (including short-circuiting) — it's really just syntax.

## 🎯 Chained Comparisons
Python lets you chain comparisons the way you'd write them on paper — `0 <= score <= 100` reads naturally and works exactly as you'd expect.

````python
score = 85
print(0 <= score <= 100)

grade = 'B'
print('A' <= grade <= 'C')
````

### 🔀 Cross-Language Note: C++ Does NOT Have This
This is a real gotcha if you're coming from C++: `0 <= score <= 100` is **legal C++ syntax**, but it does **not** mean what you'd think.

```cpp
// C++ — this COMPILES but is almost certainly a bug!
int score = 150;  // clearly out of range
if (0 <= score <= 100) {
    cout << "In range!" << endl;  // ...this prints anyway!
}
```

Why? C++ evaluates left to right: `(0 <= score)` produces a `bool`, which gets silently converted to `0` or `1` (an `int`). Then that `0`/`1` gets compared to `<= 100`, which is **always true**. The whole expression is always `true`, no matter what `score` is.

The correct C++ version has to use `&&` explicitly:
```cpp
if (0 <= score && score <= 100) { ... }
```
Python's chained comparisons are a real language feature that actually does the right thing — not just shorthand.

## 🧩 Match Statement (Preview)
Since Python 3.10, there's another way to write a long chain of "is it this specific value, or this one, or this one..." checks: `match`/`case`. You are **not** required to use this in the course, but you should recognize it when you see it.

If you know C++'s `switch`, this is the same idea, but more flexible — Python's `match` can work on strings (not just integers/enums) and can match multiple values in one `case` using `|`.

````python
day = 'Sat'

match day:
    case 'Mon' | 'Tue' | 'Wed' | 'Thu' | 'Fri':
        print('Weekday')
    case 'Sat' | 'Sun':
        print('Weekend')
    case _:
        print('Not a valid day')
````

## 🔥 Challenge (Optional) — Conditional Expressions
Python has a one-line if/else called a **conditional expression** (or "ternary"):
```python
age = 20
status = 'adult' if age >= 18 else 'minor'
print(status)
```
If you know C++, this is the same idea as `?:`, just reordered:
```cpp
// C++ — condition comes first, same as Python
string status = (age >= 18) ? "adult" : "minor";
```
Python reads more like English (value-first), while C++ reads more like the underlying logic (condition-first, then the two outcomes).
