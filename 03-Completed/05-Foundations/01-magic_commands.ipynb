## 1. Line Magics, Cell Magics, and Shell Commands
There are three kinds of "special" commands you can put in a code cell:

| Prefix | Name | Applies to | Example |
|---|---|---|---|
| `%` | Line magic | Just that one line | `%pwd` |
| `%%` | Cell magic | The **entire cell** (must be the very first line) | `%%writefile hello.py` |
| `!` | Shell command | Sends the line to your operating system's shell | `!ls` |

`%ls` is a built-in IPython magic. `!ls` actually runs the `ls` program in your shell (on Windows, use `!dir`). They often look the same, but `!` can run **any** shell command.

```python
!echo 'Hello from the shell'
```

Two tricks make `!` really useful:
- **Capture** the output into a Python variable: `files = !ls` gives you back a list of lines.
- **Pass** a Python variable into a shell command by wrapping it in `{curly_braces}`.

```python
files = !ls
print(len(files), 'items in this folder')
files[:5]
```

```python
name = 'Graylian'
!echo "Hello, {name}" 
```

## 4. Debugging and Introspection
The cell below errors **on purpose**. Read the traceback, then run `%debug` in the next cell to open the debugger at the point where it failed.

Inside the debugger (the `ipdb>` prompt), try:

| Command | What it does |
|---|---|
| `p x` | Print the value of variable `x` |
| `u` / `d` | Move **u**p / **d**own the call stack |
| `l` | List the source code around the current line |
| `q` | **Quit** the debugger. Always do this, or the cell keeps running! |

```python
def ratio(x, y):
    return x / y

# ratio(10, 0)
```

```python
%debug
```

🔁 **Before you Restart & Run All:** Run All stops at the first error, so this deliberate error would keep every cell after it from running (see notebook 02). Once you've finished exploring with `%debug`, change `ratio(10, 0)` to `ratio(10, 2)`, or comment the line out (`Ctrl+/`, or `Cmd+/` on Mac). With no error to inspect, `%debug` just prints *"No traceback has been produced"*, which is fine.

💡 `%pdb on` opens the debugger **automatically** whenever an error happens, so you don't have to type `%debug` each time. Turn it back off with `%pdb off`.

## 5. Writing and Running Files
`%%writefile` (a **cell** magic, note the double `%%`) saves the entire cell's contents to a file instead of running it as Python.

```python
%%writefile hello.py
print('Hello from a file!')
```

```python
%run hello.py
```

A few related magics:
- `%pycat hello.py` **shows** a file's contents with syntax highlighting, without running it.
- `%load hello.py` **pastes** the file's contents into the cell so you can edit it. Run the cell once to load the code, then run it again to execute it.
- `%run -i script.py` runs the script **inside** your notebook's namespace, so it can see variables you've already defined. A plain `%run` gives the script a clean slate.

```python
%pycat hello.py
```

```python
%%writefile uses_notebook_var.py
print(f'The script sees: {greeting}')
```

```python
greeting = 'hi from the notebook'
%run -i uses_notebook_var.py
```

### Auto-reloading Your Own Modules
When you `import` your own `.py` file, Python caches it. If you then edit the file, the notebook **keeps using the old version** until you restart the kernel. The `autoreload` extension fixes this:

```python
%load_ext autoreload
%autoreload 2
```

```python
%%writefile helpers.py
def greet(name):
    return f'Hello, {name}!'
```

```python
import helpers
helpers.greet('Graylian')
```

Now change the file and call the function again. You don't need to re-import or restart:

```python
%%writefile helpers.py
def greet(name):
    return f'Greetings and salutations, {name}!'
```

```python
helpers.greet('Graylian')
```

## 6. Installing Packages: `%pip` vs `!pip`
Inside a notebook, always use **`%pip install package_name`**. `%pip` installs into the same Python environment your notebook's kernel is running in.

`!pip` uses whichever `pip` your shell finds first, which may belong to a **different** Python. Then the install "works", but `import` still fails with *ModuleNotFoundError*.

The cell below only *shows* info about a package that's already installed. It doesn't install anything:

```python
%pip show numpy
```

## 📦 Collect Your Output Files
This notebook created several `.py` files. Run the cell below **last**. It moves them into a `01-OutputFiles` folder so they're easy to find and grade. Upload that folder along with this notebook.

It's safe to run more than once: files that are already in the folder get replaced with the newest version.

```python
from pathlib import Path

out_dir = Path('01-OutputFiles')
out_dir.mkdir(exist_ok=True)

for name in ['hello.py', 'uses_notebook_var.py', 'helpers.py', 'my_script.py']:
    f = Path(name)
    if f.exists():
        f.replace(out_dir / name)
        print(f'Moved {name} -> {out_dir}/')
    else:
        print(f'Not found (skipped): {name}')
```
