````python
import pandas as pd
````

## 📥 Load a CSV
This module's folder includes `titanic.csv` — the classic Titanic passenger dataset — so everything here works offline, with no download step.

````python
df = pd.read_csv('titanic.csv')
df.head()
````

## 📊 Peek at the Data

````python
print('First 5 rows:')
print(df.head())

print('\nLast 5 rows:')
print(df.tail())
````

## 🧠 Understand the Structure

````python
print('Shape:', df.shape)
print('\nColumns:', df.columns.tolist())
````

`(891, 12)` means **891 rows and 12 columns**. `.columns` gives you an Index object, and `.tolist()` turns it into a plain Python list.

Each column holds one **data type** (dtype):

````python
df.dtypes
````

| dtype | Holds | Titanic examples |
|---|---|---|
| `int64` | Whole numbers | `Survived`, `Pclass`, `SibSp` |
| `float64` | Decimals, **or** whole numbers with missing values (`NaN` is a float) | `Age`, `Fare` |
| `object` / `str` | Text (`str` in pandas 3, `object` in older versions) | `Name`, `Sex`, `Ticket` |
| `bool` | `True` / `False` | none here |
| `datetime64` | Dates and times | none here (see notebook 04) |

💡 `Age` holds whole numbers, so why is it `float64`? It has missing values, and `NaN` only exists as a float. Notice that, because it's a clue there's something to clean.

## 🧼 Data Types and Missing Values

````python
# Rows, columns, non-null counts per column, dtypes, and memory use
print('Info:')
df.info()

print('\nMissing values per column:')
print(df.isnull().sum())
````

## 📈 Summary Statistics

````python
df.describe(include='all')
````

By itself, `.describe()` only summarizes the **number** columns (count, mean, std, min, quartiles, max). `include='all'` adds the text columns too: their `count`, number of `unique` values, the most common value (`top`), and how often it appears (`freq`).

## ✏️ Renaming Columns

````python
# Rename 'Sex' to 'Gender' for clarity
df.rename(columns={'Sex': 'Gender'}, inplace=True)
df.head(2)
````

## 🔎 Duplicates and Distinct Values
Two more quick questions to ask of any new dataset: **are any rows repeated**, and **what values does a column actually contain?**

````python
print('Duplicate rows:', df.duplicated().sum())
print('Ports:', df['Embarked'].unique())          # every distinct value
print('How many ports:', df['Embarked'].nunique())  # how many distinct (NaN not counted)
````

````python
# How often each value appears. dropna=False also counts the missing ones
df['Embarked'].value_counts(dropna=False)
````

Notebook 04 shows how to remove duplicates, and notebook 07 does much more with `value_counts()`.

## ✅ A First-Look Checklist
Whenever you get an unfamiliar dataset, run these before anything else:

```python
df.head()
df.shape
df.columns.tolist()
df.info()
df.describe(include='all')
df.isnull().sum()
df.duplicated().sum()
```

Then answer these questions **in words**:
1. **What does one row represent?** (One passenger? One sale? One day?)
2. Which column(s) **identify** each row?
3. Which columns are **numbers**, which are **categories**, and which are **dates**?
4. Are there **missing values** or **duplicates**?
5. Do the values look **reasonable**? (Ages from 0 to 80? Fares that are never negative?)

> **The core habit:** inspect first, select carefully, and prefer whole-column operations over row-by-row loops.
